
## 8. Data Preprocessing


```python

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_encoded
)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)

print("\nTraining class distribution:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nTesting class distribution:")
print(pd.Series(y_test).value_counts().sort_index())

print("\nClass labels:")
print(label_encoder.classes_)
```


## 9. Feature Scaling




```python

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully!")

print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)

print("\nTraining data mean:", np.mean(X_train_scaled).round(4))
print("Training data standard deviation:", np.std(X_train_scaled).round(4))
```


## 10. Kernel Functions


```python

gamma_value = 1 / X_train_scaled.shape[1]

kernel_functions = {
    "Linear": linear_kernel,
    "Polynomial": polynomial_kernel,
    "RBF": rbf_kernel
}

K_train = {
    "Linear": linear_kernel(X_train_scaled, X_train_scaled),

    "Polynomial": polynomial_kernel(
        X_train_scaled,
        X_train_scaled,
        degree=3,
        gamma=gamma_value,
        coef0=1
    ),

    "RBF": rbf_kernel(
        X_train_scaled,
        X_train_scaled,
        gamma=gamma_value
    )
}

K_test = {
    "Linear": linear_kernel(X_test_scaled, X_train_scaled),

    "Polynomial": polynomial_kernel(
        X_test_scaled,
        X_train_scaled,
        degree=3,
        gamma=gamma_value,
        coef0=1
    ),

    "RBF": rbf_kernel(
        X_test_scaled,
        X_train_scaled,
        gamma=gamma_value
    )
}

for kernel_name in K_train:
    print(f"{kernel_name} Kernel:")
    print("Training matrix:", K_train[kernel_name].shape)
    print("Testing matrix:", K_test[kernel_name].shape)
    print("-" * 40)
```


## 11. Kernel Feature Preparation


```python

kernel_scalers = {}
K_train_scaled = {}
K_test_scaled = {}

for name in K_train:

    kernel_scalers[name] = StandardScaler()

    K_train_scaled[name] = kernel_scalers[name].fit_transform(
        K_train[name]
    )

    K_test_scaled[name] = kernel_scalers[name].transform(
        K_test[name]
    )

    print(f"{name} Kernel:")
    print("Training:", K_train_scaled[name].shape)
    print("Testing:", K_test_scaled[name].shape)
    print("-" * 40)
```


## 12.  RMSprop Optimizer


```python
def compute_loss(X, y, W, b, C):
    n_samples = X.shape[0]
    n_classes = W.shape[1]

    Y = -np.ones((n_samples, n_classes))
    Y[np.arange(n_samples), y] = 1

    margins = np.maximum(0, 1 - Y * (X @ W + b))

    return 0.5 * np.sum(W ** 2) + (C / n_samples) * np.sum(margins ** 2)


def train_rmsprop_svm(
    X, y,
    learning_rate=0.001,
    epochs=50,
    C=1,
    batch_size=64,
    rho=0.9,
    epsilon=1e-8,
    random_state=42
):
    rng = np.random.default_rng(random_state)

    n_samples, n_features = X.shape
    n_classes = len(np.unique(y))

    Y = -np.ones((n_samples, n_classes))
    Y[np.arange(n_samples), y] = 1

    W = np.zeros((n_features, n_classes))
    b = np.zeros(n_classes)

    cache_W = np.zeros_like(W)
    cache_b = np.zeros_like(b)

    loss_history = []

    for epoch in range(epochs):

        indices = rng.permutation(n_samples)

        for start in range(0, n_samples, batch_size):

            batch_idx = indices[start:start + batch_size]

            X_batch = X[batch_idx]
            Y_batch = Y[batch_idx]

            scores = X_batch @ W + b
            margins = np.maximum(0, 1 - Y_batch * scores)
            error = Y_batch * margins

            gradient_W = W - (
                2 * C * (X_batch.T @ error) / len(batch_idx)
            )

            gradient_b = -2 * C * np.mean(error, axis=0)

            cache_W = (
                rho * cache_W
                + (1 - rho) * gradient_W ** 2
            )

            cache_b = (
                rho * cache_b
                + (1 - rho) * gradient_b ** 2
            )

            W -= learning_rate * gradient_W / (
                np.sqrt(cache_W) + epsilon
            )

            b -= learning_rate * gradient_b / (
                np.sqrt(cache_b) + epsilon
            )

        loss = compute_loss(X, y, W, b, C)
        loss_history.append(loss)

        if not np.isfinite(loss):
            break

    return W, b, loss_history
```

## 12.1 Hyperparameter Tuning (Validation Split)



```python
fit_idx, val_idx = train_test_split(
    np.arange(len(y_train)),
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_train
)

param_grid = {
    "C": [0.1, 1, 10, 100],
    "learning_rate": [0.0001, 0.001, 0.01, 0.1]
}

TUNE_EPOCHS = 30

tuning_results = []

for kernel_name in K_train_scaled:

    K_all = K_train_scaled[kernel_name]

    for C in param_grid["C"]:
        for lr in param_grid["learning_rate"]:

            W, b, loss_history = train_rmsprop_svm(
                K_all[fit_idx],
                y_train[fit_idx],
                learning_rate=lr,
                epochs=TUNE_EPOCHS,
                C=C
            )

            if np.isfinite(loss_history[-1]):

                val_pred = np.argmax(
                    K_all[val_idx] @ W + b,
                    axis=1
                )

                val_f1 = f1_score(
                    y_train[val_idx],
                    val_pred,
                    average="macro",
                    zero_division=0
                )

                val_acc = accuracy_score(
                    y_train[val_idx],
                    val_pred
                )

            else:
                val_f1, val_acc = np.nan, np.nan

            tuning_results.append({
                "Kernel": kernel_name,
                "C": C,
                "Learning Rate": lr,
                "Final Loss": loss_history[-1],
                "Val Accuracy": val_acc,
                "Val F1-Score": val_f1
            })

tuning_results = pd.DataFrame(tuning_results)

best_params = {}

for kernel_name in K_train_scaled:

    subset = tuning_results[
        tuning_results["Kernel"] == kernel_name
    ]

    best_row = subset.loc[
        subset["Val F1-Score"].idxmax()
    ]

    best_params[kernel_name] = {
        "C": best_row["C"],
        "learning_rate": best_row["Learning Rate"]
    }

print("Best RMSprop Hyperparameters:")
print("=" * 50)

for kernel_name, params in best_params.items():
    print(
        f"{kernel_name}: C={params['C']}, "
        f"learning_rate={params['learning_rate']}"
    )

display(
    tuning_results.sort_values(
        by=["Kernel", "Val F1-Score"],
        ascending=[True, False]
    ).reset_index(drop=True)
)
```


## 13. Model Training



```python
FINAL_EPOCHS = 100

rmsprop_models = {}
rmsprop_training_results = []

for kernel_name in K_train_scaled:

    params = best_params[kernel_name]

    print(
        f"Training RMSprop with {kernel_name} Kernel "
        f"(C={params['C']}, lr={params['learning_rate']})..."
    )

    start_time = time.time()

    W, b, loss_history = train_rmsprop_svm(
        K_train_scaled[kernel_name],
        y_train,
        learning_rate=params["learning_rate"],
        epochs=FINAL_EPOCHS,
        C=params["C"]
    )

    training_time = time.time() - start_time

    rmsprop_models[kernel_name] = {
        "W": W,
        "b": b,
        "loss_history": loss_history,
        "C": params["C"]
    }

    rmsprop_training_results.append({
        "Kernel": kernel_name,
        "C": params["C"],
        "Learning Rate": params["learning_rate"],
        "Training Time (s)": round(training_time, 2),
        "Initial Loss": loss_history[0],
        "Final Loss": loss_history[-1],
        "Epochs": len(loss_history)
    })

    print(
        "Done in", round(training_time, 2),
        "seconds | Final loss:", round(loss_history[-1], 4)
    )
    print("-" * 40)

rmsprop_training_results = pd.DataFrame(
    rmsprop_training_results
)

display(rmsprop_training_results)
```


## 14. Loss Visualization



```python
plt.figure(figsize=(10, 6))

for kernel_name, model in rmsprop_models.items():
    plt.plot(
        model["loss_history"],
        label=kernel_name
    )

plt.title("RMSprop Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.tight_layout()
plt.show()
```


## 15. Model Evaluation


```python
rmsprop_predictions = {}
rmsprop_evaluation_results = []

for kernel_name in K_test_scaled:

    model = rmsprop_models[kernel_name]

    scores = (
        K_test_scaled[kernel_name] @ model["W"]
        + model["b"]
    )

    predictions = np.argmax(scores, axis=1)

    rmsprop_predictions[kernel_name] = predictions

    rmsprop_evaluation_results.append({
        "Kernel": kernel_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions,
            average="macro",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions,
            average="macro",
            zero_division=0
        ),
        "F1-Score": f1_score(
            y_test, predictions,
            average="macro",
            zero_division=0
        ),
        "Train Loss": model["loss_history"][-1],
        "Test Loss": compute_loss(
            K_test_scaled[kernel_name],
            y_test,
            model["W"],
            model["b"],
            model["C"]
        )
    })

rmsprop_evaluation = pd.DataFrame(
    rmsprop_evaluation_results
)

display(
    rmsprop_evaluation.sort_values(
        by="Accuracy",
        ascending=False
    ).reset_index(drop=True)
)
```


## 16. Classification Reports


```python
for kernel_name, predictions in rmsprop_predictions.items():

    print(f"\n{kernel_name} Kernel Classification Report")
    print("=" * 60)

    print(classification_report(
        y_test,
        predictions,
        labels=np.arange(len(label_encoder.classes_)),
        target_names=label_encoder.classes_,
        zero_division=0
    ))
```
